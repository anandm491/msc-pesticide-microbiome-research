## 6. Descriptive statistics

This section calculates the count, mean, standard deviation, minimum, median, and maximum for each soil property within each management group.

These are **descriptive statistics only** and should not be interpreted as evidence of statistically significant treatment differences.


````python
available_properties = [
    col for col in SOIL_PROPERTIES.values()
    if col in df.columns
]

if GROUP_COLUMN in df.columns and available_properties:
    descriptive_stats = (
        df.groupby(GROUP_COLUMN)[available_properties]
          .agg(["count", "mean", "std", "min", "median", "max"])
          .round(3)
    )
    display(descriptive_stats)
````

## 7. Create box plots

The box plots below compare soil properties among the five Glenlea management/rotation groups.

Individual observations are also shown because sample sizes are small.


````python
GROUP_ORDER = [
    "1-Organic",
    "16-Organic",
    "1-Conventional",
    "16-Conventional",
    "Native grass",
]

def soil_boxplot(data, group_col, value_col, y_label, title):
    groups = []
    labels = []

    for group in GROUP_ORDER:
        values = data.loc[data[group_col] == group, value_col].dropna()
        if len(values) > 0:
            groups.append(values.values)
            labels.append(f"{group}\n(n={len(values)})")

    fig, ax = plt.subplots(figsize=(10, 6))
    ax.boxplot(groups, labels=labels, showmeans=True)

    # Show individual observations.
    for i, values in enumerate(groups, start=1):
        ax.scatter([i] * len(values), values, zorder=3)

    ax.set_title(title)
    ax.set_ylabel(y_label)
    ax.set_xlabel("")
    ax.tick_params(axis="x", rotation=0)
    fig.tight_layout()
    plt.show()
````

### 7.1 Total Organic Carbon


````python
col = SOIL_PROPERTIES["Total Organic Carbon (%)"]
if GROUP_COLUMN in df.columns and col in df.columns:
    soil_boxplot(
        df, GROUP_COLUMN, col,
        "Total Organic Carbon (%)",
        "Total Organic Carbon"
    )
````

### 7.2 Organic Matter


````python
col = SOIL_PROPERTIES["Organic Matter (%)"]
if GROUP_COLUMN in df.columns and col in df.columns:
    soil_boxplot(
        df, GROUP_COLUMN, col,
        "Organic Matter (%)",
        "Organic Matter"
    )
````
